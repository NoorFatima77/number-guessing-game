```python
import random
import ipywidgets as widgets
from IPython.display import display, clear_output

def play_guessing_game():
    # State variables
    game_state = {
        "secret_number": 0,
        "max_range": 100,
        "score": 100,
        "attempts": 0,
        "game_active": False
    }

    # UI Components
    title_label = widgets.HTML(value="<h2 style='color: #2b580c;'>🎯 Number Guessing Mini-Game</h2>")

    range_dropdown = widgets.Dropdown(
        options=[('Easy (1 to 10)', 10), ('Hard (1 to 100)', 100)],
        value=100,
        description='Select Range:',
        style={'description_width': 'initial'}
    )

    start_button = widgets.Button(
        description='Start Game',
        button_style='success',
        icon='play'
    )

    guess_text = widgets.IntText(
        value=0,
        description='Your Guess:',
        disabled=True,
        style={'description_width': 'initial'}
    )

    submit_button = widgets.Button(
        description='Submit Guess',
        button_style='primary',
        disabled=True,
        icon='check'
    )

    output_area = widgets.Output(layout={'border': '1px solid #ccc', 'padding': '10px', 'margin-top': '10px'})

    def start_game(b):
        game_state["max_range"] = range_dropdown.value
        game_state["secret_number"] = random.randint(1, game_state["max_range"])
        game_state["score"] = 100
        game_state["attempts"] = 0
        game_state["game_active"] = True

        guess_text.disabled = False
        submit_button.disabled = False

        with output_area:
            clear_output()
            print(f"🚀 Game Started! I have selected a number between 1 and {game_state['max_range']}.")
            print(f"Starting Score: {game_state['score']} points. Make your first guess below!\n")

    def handle_guess(b):
        if not game_state["game_active"]:
            return

        user_guess = guess_text.value
        game_state["attempts"] += 1

        with output_area:
            # 1. Validate if number is within range
            if user_guess < 1 or user_guess > game_state["max_range"]:
                print(f"⚠️ Please enter a valid number between 1 and {game_state['max_range']}.")
                return

            # 2. Compare numbers and compute difference
            diff = abs(game_state["secret_number"] - user_guess)

            if user_guess == game_state["secret_number"]:
                print(f"\n🎉 Congratulations! You guessed the correct number ({game_state['secret_number']})!")
                print(f"⭐ Final Score: {max(0, game_state['score'])} | Total Attempts: {game_state['attempts']}")
                game_state["game_active"] = False
                guess_text.disabled = True
                submit_button.disabled = True
            else:
                game_state["score"] -= 15
                print(f"❌ Wrong guess! (Difference: {diff}) | Current Score: {max(0, game_state['score'])}")

                # 3. Provide Dynamic Clues (Greater/Smaller & Divisibility)
                if user_guess < game_state["secret_number"]:
                    print("💡 Hint: The secret number is greater than your guess.")
                else:
                    print("💡 Hint: The secret number is smaller than your guess.")

                divisible_by = [i for i in range(2, 10) if game_state["secret_number"] % i == 0]
                if divisible_by:
                    print(f"💡 Hint: The secret number is divisible by {random.choice(divisible_by)}.")
                print("-" * 40)

                if game_state["score"] <= 0:
                    print(f"\n💀 Game Over! You've run out of points. The secret number was {game_state['secret_number']}.")
                    game_state["game_active"] = False
                    guess_text.disabled = True
                    submit_button.disabled = True

    # Bind button events
    start_button.on_click(start_game)
    submit_button.on_click(handle_guess)

    # Display UI layout in Google Colab
    ui_layout = widgets.VBox([
        title_label,
        widgets.HBox([range_dropdown, start_button]),
        widgets.HBox([guess_text, submit_button]),
        output_area
    ])

    display(ui_layout)

# Run the interactive game interface
play_guessing_game()
```
